```python
5.	Explain your understanding of overfitting and underfitting in detail. How do you identify and measure overfitting? 

Data science is all about model training and predictions.

We train models where we give the problem statment and also results.

They learn the problem statement and try to produce the algorithm that will help us to produce minimal loss and close to the actual value

Since we produce the actual y value or predictor value, the algorithm continously works to produce the best results.

After learning, we give the images/data and expects the algorithm  to predict the results

However, in this process we may incur overfitting and underfitting

overfitting - low bias and high variance

            The algorithm gives best results in training with min loss(low bias) and high variance, less accuracy in the testing phase. This occurs because the algorithm overlearns

  the data and be able to give best results but does provide an optimal algorithm that also performs best in testing. When testing occurs, the overfitting model is unable to produce

  good accuracy thereby making the model poor.

Underfitting: High  bias and low variance

Underfitting is when the model built itself is poor and causes high loss and low accuracy. The model has not learnt and unable to perform well even in the training phase.

Due to poor learning, the model also performs poor in the testing phase too and thereby we classify the model as underfitting  

```

```python
6.	What is the use of batch size and why is it used in Deep Learning?

Batch size is one of the hyper parameter tuning techniques. It implies the number of images that we pass on one round/epoch. 

Suppose we have 60,000 images and instead of feeding all the 60000 images at the same time which brings in conflict of handling different weights/biases and different 
output probabilities , we introuduce a batch size. Batch size=32 is default causing only 32 images per patch causing only 32 images to be dealt by the neural network at a time which
may significanly reduce the overhead of RAM and increase the computing efficiency also.

However, it may increase the time but it may not have a significant difference for a average batch size

High batch size may be blow up the RAM and very low batch size like 1 , 2.. < 16 will not effective contribute to repetetive learning of the model.
```
