```python
import tensorflow as tf
import pandas as pd
```

```python
df_wine=pd.read_csv("wine.data",header=None)
```

```python
df_wine.columns=['Target','Alcohol','Malic_acid','Ash','Alcalinity','Magnesium','Total_phenols',
                                                    'Flavanoids','Nonflavanoid','Proanthocyanins','Color_intensity','Hue',
                                                    'diluted_wines','Proline']
```

```python
df_wine['Target'].value_counts()
```

```python
from sklearn.preprocessing import StandardScaler 
```

```python
x_train=df_wine.iloc[:,1:]
y_train=df_wine.iloc[:,0]
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit(x_train)
xtrain=scaler.transform(x_train)
```

```python
x_train=pd.DataFrame(xtrain)
```

```python
from sklearn.model_selection import train_test_split
xtr,xte,ytr,yte=train_test_split(x_train,y_train,test_size=0.2,random_state=1)
```

```python
ytr.shape
```

```python
del model
model=tf.keras.models.Sequential()
model.add(tf.keras.layers.Input(shape=(13),name='Input'))
model.add(tf.keras.layers.Dense(units=64,activation=tf.keras.activations.relu,name='Hidden'))
model.add(tf.keras.layers.Dense(units=3,activation=tf.keras.activations.softmax,name='output'))
```

```python
model.summary()
```

```python
model.compile(optimizer=tf.keras.optimizers.SGD(),loss=tf.keras.losses.categorical_crossentropy,metrics=['acc'])
#y_train = tf.keras.utils.to_categorical(y_train)
model.fit(xtr,ytr,validation_split=0.1,epochs=10)
```
