```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
def sigmoid(x1):
    ypred1 = 1 / (1+ math.exp(-x1))
    return ypred1 

def tanh(x1):
    ypred1 = (math.exp(x1) - math.exp(-x1))/ (math.exp(x1) + math.exp(-x1))
    return ypred1 


def relu(x1):
    if x1 > 0:
      return x1
    else:
      return 0

def RunThis(alpha1,epoch,activation):
    y=10
    x=5
    w=np.random.rand()
    b=np.random.rand()
    wlist=[]
    blist=[]
    errorlist=[]
    ypredlist=[]
    results=pd.DataFrame()
    #-Learning rate
    for i in range(epoch):
        if activation == 'sigmoid':
          ypred= round(w * sigmoid(x) + b,4)
        elif activation == 'tanh':
          ypred= round(w * tanh(x) + b,4)
        else:
          ypred= round(w * relu(x) + b,4)
        error = round((y-ypred)**2,4)
        wlist.append(w)
        blist.append(b)
        ypredlist.append(ypred)
        errorlist.append(error)
        dedw= -2 * x * (y-ypred)
        dedb= -2 * (y-ypred)
        w= w-alpha1 * dedw
        b= b-alpha1 * dedb
    df_results = pd.DataFrame({"X":x,"Y":y,"Ypred":[ypred],
                               "Alpha":[alpha1],"Epoch" :[epoch],"Activation":[activation],"Error" :[error]})
    return df_results
```

```python
epoch = [1,5,10,20]
LR=[0.001,0.01,0.1,1]
Activation= ['sigmoid','tanh','Relu']


results=pd.DataFrame()
for i in range(0,len(epoch)):
  for j in range(0,len(Activation)):
    for k in range(0,len(LR)):
      final_result = RunThis(LR[k],epoch[i],Activation[j])
      results=pd.concat([results,final_result])

```

```python
results.sort_values(by=['Activation','Epoch','Alpha'],inplace=True)
#results.reset_index()
```

```python
# Assignment 6

6.	Mathematically, derive the complete process of a single perceptron from an input to output and perform the below task:
a.	Explain the above in 3 phases of Feed Forward, Error Calculation, Back Propagation mathematically.
b.	Explain the role and importance of each and every parameter other than x and y.

Every problem in this world is converted into a linear equation.

Neurons are capable of performing solving problems only in linear fashion 

For single perceptron , the equation is 

 y = w * x+b

 where x is th input and y is the output

 In training phase, I have my x - ground truth and also y which is the actual output. The model is now trained very well to determine the best value of w and b

 This is a single perceptron and hence only 1 weight and 1 bias.

Feed forward:

Initially, the weights and bias are random.

Now we calculate ypred = w * x + b

Error calculation:

We find the difference in the squared errors

E= Square of (y-ypred)

We have the error and component and we need to adjust weights and bias to minimise the error so that y = ypred

To do that we differential error with respect to w and b

e = square of (y-w1 * x - b)

de/dw = = 2(y- ypred)(0-x-0)
        -2 * x * (y-ypred)
de/db = 2 * (y-ypred)(0-0-1)
      = - 2 * (y-ypred)
Back propagation:

After calculating the error with respect to w and b , these neuron will now accept new weights and bias

w(new) = w(old) - de/dw
b(new) = b(old) - de/db

This is again fed into the linear equation of y= w * x + b and new ypred is determined

The process will be repeated until the minimum loss i.e error is minimum and ypred = y.



```
