```python
import cv2
import matplotlib.pyplot as plt
```

```python
def print_image(img):
  image=cv2.imread(img)
  image=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
  print(image.shape)
  plt.imshow(image)
```

```python

folder = "train/"
images = []
labels = []
imagenames=[]
height=[]
weight=[]
dim=[]
for _imageName in os.listdir("train"):
  imageName = folder + _imageName
  imagenames.append(imageName)
  image = cv2.imread(imageName)
  image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
  #image = cv2.resize(image, (150, 150))
  height.append(image.shape[0]) 
  weight.append(image.shape[1]) 
  dim.append(image.shape[2])
  images.append(image)
  if _imageName.lower().startswith("cat"):
    labels.append(0)
  else:
    labels.append(1)
```

```python
print(labels)
```

```python
import tensorflow as tf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
idg=ImageDataGenerator(rescale=1/255.0,validation_split=0.1)
train_gen=idg.flow_from_directory(directory = "train_data",target_size=(150,150),batch_size=64,shuffle=True,subset="training")
val_gen=idg.flow_from_directory(directory="train_data",target_size=(150,150,),batch_size=64,shuffle=True,subset="validation")
```

```python
model=tf.keras.models.Sequential()
model.add(tf.keras.layers.Input(shape=(150,150,3),name='input'))
model.add(tf.keras.layers.Flatten())
model.add(tf.keras.layers.Dense(activation=tf.keras.activations.relu,units=150,name='Dense1'))
model.add(tf.keras.layers.Dense(activation=tf.keras.activations.softmax,units=2,name='Outlayer'))
```

```python
model.summary()
```

```python
model.compile(optimizer=tf.keras.optimizers.SGD(),loss=tf.keras.losses.CategoricalCrossentropy(),metrics=['acc'])
```

```python
model.fit(train_gen,validation_data=val_gen,epochs=10)
```
